```python
import numpy as np

# Create two matrices
A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])

# Display matrices
print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)

# 1. Matrix Addition
addition = A + B
print("\nMatrix Addition:")
print(addition)

# 2. Matrix Subtraction
subtraction = A - B
print("\nMatrix Subtraction:")
print(subtraction)

# 3. Matrix Multiplication
multiplication = np.dot(A, B)
print("\nMatrix Multiplication:")
print(multiplication)

# 4. Transpose of Matrix A
transpose_A = A.T
print("\nTranspose of Matrix A:")
print(transpose_A)

# 5. Transpose of Matrix B
transpose_B = B.T
print("\nTranspose of Matrix B:")
print(transpose_B)
```
